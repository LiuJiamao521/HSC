# msigdb 34837 gene sets -- all time

```r
library(Seurat)
library(UCell)
library(clusterProfiler)
library(foreach)
library(doParallel)

# Define input and output directories
input_dir <- "/cluster2/huanglab/jiamao/Project/HumanSpinalCord/Work/P3/ST/data/CalculateCells/Smooth/RDS/"
output_dir <- "/cluster2/huanglab/jiamao/Project/HumanSpinalCord/Work/P3/ST/Ucell/msigdb/"

# Define stage mapping
file_mapping <- list(
  "gw7-1" = "gw71",
  "gw7-2" = "gw72",
  "gw7-3" = "gw73",
  "gw6" = "gw6"
  # "gw9" = "gw9",
  # "gw10" = "gw10",
  # "gw12" = "gw12",
  # "gw16" = "gw16",
  # "gw17-1" = "gw171",
  # "gw17-2" = "gw172"
)

# Load MSigDB gene sets; 34 indicates all gene sets
MSigDB <- '/cluster2/huanglab/jiamao/Apps/MSigDB/msigdb_v2024.1.Hs_GMTs'
gmtfs <- list.files(MSigDB, pattern = 'symbols.gmt')
geneset <- clusterProfiler::read.gmt(gmtfile = file.path(MSigDB, gmtfs[34]))
geneset <- split(geneset$gene, geneset$term)

# Set number of parallel cores (adjust by available resources)
n_cores <- 10  # Recommended: physical cores minus 2 to reserve system resources
cl <- makeCluster(n_cores)
registerDoParallel(cl)

# Process each sample in parallel
foreach(input_key = names(file_mapping), 
        .packages = c("Seurat", "UCell", "Matrix"), # Packages to load explicitly
        .export = c("input_dir", "output_dir", "file_mapping", "geneset") # Variables to export
) %dopar% {
  
  output_key <- file_mapping[[input_key]]
  
  # Generate file names dynamically
  input_file <- paste0("SeuratObj_", input_key, ".rds")
  output_adjust_file <- paste0("adjust_ucell_msigdb_matrix_", output_key, ".csv")
  output_file <- paste0("ucell_msigdb_matrix_", output_key, ".csv")

  # Read data
  obj <- readRDS(file.path(input_dir, input_file))
  
  # Extract expression matrix
  exp <- as.matrix(obj@assays$RNA@counts)
  
  # Compute UCell scores
  scores <- UCell::ScoreSignatures_UCell(exp, features = geneset, maxRank = 2000)
  
  #------------adjust----------------
  # Normalize score matrix
  nFeature <- obj$nFeature_RNA
  scores_normalized <- sweep(
    x = scores,
    MARGIN = 1,
    STATS = nFeature,
    FUN = "/"
  )
  
  # Save results
  write.csv(scores_normalized, file.path(output_dir, output_adjust_file))
  write.csv(scores, file.path(output_dir, output_file))
  
  # Return progress message
  return(paste("Processed", input_file, "->", output_file))
} -> results  # Collect all returned messages

# Stop cluster
stopCluster(cl)
```

```r
rm(obj, exp, scores)
gc()
```
