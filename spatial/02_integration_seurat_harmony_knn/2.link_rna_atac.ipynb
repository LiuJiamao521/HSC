```python
import pandas as pd
```

```python
cluster_df = pd.read_csv('/cluster2/huanglab/jiamao/Project/HumanSpinalCord/Work/P3/ST/multimapping/GW16/cluster_df_add.atac.txt', sep='\t', index_col=0)
cluster_df = cluster_df.sort_index()
cluster_df.head()
```

```python
rna_anno = pd.read_csv('/cluster2/huanglab/jiamao/Project/HumanSpinalCord/Work/P3/ST/multimapping/GW16/rna_anno.csv', index_col=0)
rna_anno.head()
```

```python
common_indices = cluster_df.index.intersection(rna_anno.index)

link_atac_matrix = pd.DataFrame({
    'id_atac': common_indices,
    'id_rna': common_indices,
    'cluster_rna': rna_anno.loc[common_indices, 'cell_name']
})
link_atac_matrix.head()
```

```python
link_atac_matrix.to_csv("/cluster2/huanglab/jiamao/Project/HumanSpinalCord/Work/P3/ST/multimapping/GW16/link_atac_matrix.csv", index=False)
```
