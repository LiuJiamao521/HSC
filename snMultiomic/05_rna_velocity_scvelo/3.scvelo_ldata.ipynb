```python
import glob
import os

samples = [
    "GW7", "GW12", "GW6", "GW9", "GW10", "GW16", "GW17", "GW20"
]

# Path to loom file
file_paths = [
    glob.glob(f"/cluster/huanglab/lab/LabDataset/jiamao/HumanSpinalCord/ProcessData/Cellranger/{sample}/velocyto/*.loom")[0]
    for sample in samples[:2]
] + [
    glob.glob(f"/cluster/huanglab/lab/LabDataset/jiamao/HumanSpinalCord/ProcessData/Cellranger/{sample}/*.loom")[0]
    for sample in samples[2:]
]
```

```python
file_paths
```

```python
ldatas = []
for i, path in enumerate(file_paths):
    ldata = scv.read(path, cache=True)
    ldata.obs.index = ldata.obs.index.str.replace(r'^.*:|x$', '', regex=True) + f'-1'
    ldata.var_names_make_unique() 
    ldatas.append(ldata)
```

```python
ldata_combined = ldatas[0].concatenate(ldatas[1:])
```

```python
ldata_combined.write('/cluster2/huanglab/jiamao/Project/Tools/RNAvelocity/1.hscLoom_raw.h5ad')
```
