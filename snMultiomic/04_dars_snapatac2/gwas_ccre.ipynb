```python
# import pandas as pd
dar = pd.read_csv('/cluster2/huanglab/jiamao/Project/HumanSpinalCord/Work/P3/ATAC/Snapatac2/HSC/hsc_marker_celltype_dar.csv')
marker_peaks = dar.groupby('cell_type')['peak'].apply(list).to_dict()

print(type(marker_peaks))
print(marker_peaks.keys())
```

```python
import os
import pandas as pd

# 1. 定义输出目录并创建
output_dir = '/cluster2/huanglab/jiamao/Project/HumanSpinalCord/Work/P3/ST/GWAS/bed'
os.makedirs(output_dir, exist_ok=True)

# 用于存储所有去重后的 peak
all_peaks_set = set()

# 2. 定义一个辅助函数：将 formatted strings (chr:start-end) 转为 DataFrame
def peaks_to_df(peak_list):
    # 将 index 或 list 转为 Series
    s = pd.Series(list(peak_list))
    # 使用正则表达式提取 chr, start, end
    # (.*?):  匹配冒号前的所有内容作为 chr
    # (\d+)-  匹配冒号后、横杠前的数字作为 start
    # (\d+)   匹配横杠后的数字作为 end
    df = s.str.extract(r'^(.*?):(\d+)-(\d+)$')
    return df

# 3. 遍历字典，生成各个细胞类型的 bed 文件
for cell_type, peaks in marker_peaks.items():
    # 收集到总集合中
    all_peaks_set.update(peaks)
    
    # 转换格式
    df_bed = peaks_to_df(peaks)
    
    # 处理文件名：防止细胞类型名称中包含 '/' (如 v0/1) 导致路径错误
    safe_name = cell_type.replace('/', '_').replace(' ', '_')
    file_path = os.path.join(output_dir, f"{safe_name}.bed")
    
    # 保存为 bed 文件 (无表头，无索引，制表符分隔)
    df_bed.to_csv(file_path, sep='\t', header=False, index=False)
    print(f"Generated: {file_path}")

# # 4. 生成总的 hsc.bed
# # 将集合转为列表
# all_peaks_list = list(all_peaks_set)
# df_all = peaks_to_df(all_peaks_list)

# # 为了规范，最好对 bed 文件进行排序 (按 chr 和 start)
# # 如果不需要排序，可以注释掉下面两行
# # 注意：bedtools 等工具通常喜欢排序过的文件
# df_all[1] = df_all[1].astype(int) # 转换 start 列为数字以便排序
# df_all = df_all.sort_values(by=[0, 1]) # 按 chr 和 start 排序

# # 保存 hsc.bed
# hsc_path = os.path.join(output_dir, ".hsc.bed")
# df_all.to_csv(hsc_path, sep='\t', header=False, index=False)
# print(f"Generated Total: {hsc_path}")
```
