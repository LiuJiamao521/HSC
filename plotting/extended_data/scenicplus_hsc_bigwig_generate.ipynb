```python
import snapatac2 as snap
import numpy as np
import pandas as pd
# Set the environment variable to the desired cache path
import os
import scanpy as sc
```

```python
hsc_peaks = sc.read_h5ad("/cluster2/huanglab/jiamao/Project/HumanSpinalCord/Work/P3/ATAC/Snapatac2/HSC/peak_mat.h5ad")
snap.ex.export_coverage(hsc_peaks,groupby="cell_type_snRNA",out_dir="/cluster2/huanglab/jiamao/Project/HumanSpinalCord/Work/P3/ATAC/Scenicplus/HSC/bigwig",output_format="bigwig")
```

```python
peak_df = hsc_peaks.var.index.to_series().str.extract(r'(chr[\w]+):(\d+)-(\d+)')
peak_df.columns = ['chrom', 'start', 'end']
peak_df.to_csv("/cluster2/huanglab/jiamao/Project/HumanSpinalCord/Work/P3/ATAC/Scenicplus/HSC/outs/consensus_peaks.bed", sep="\t", header=False, index=False)
```
